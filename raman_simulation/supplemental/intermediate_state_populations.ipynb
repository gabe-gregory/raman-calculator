```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
sys.path.insert(0, r'Z:\Users\Gregory\Ca40_multi_photon')
sys.path.insert(0, r'Z:\Users\Gregory\Ca40_multi_photon\raman_simulation')
import useful_funcs
import simulation as s
from matplotlib.colors import TwoSlopeNorm
import scienceplots
import matplotlib.colors as colors
import matplotlib.ticker as ticker
```

```python
sim = s.sim()
sim.ls_order = 2
sim.get_f()

sim.w0r = 31.52*1e-6
sim.r_sp = .0755
sim.r_pi = .0016
sim.w0b = 30.0*1e-6
sim.b_sp = .3355
sim.b_pi = .329
```

```python
P_bank = np.linspace(15, 200, 10)*1e-3
sim.power_scaling(P_bank)
```

```python
xtick_positions = [10,100]
xtick_labels = ['10','100']
s1 = 1.5
s2 = 1.5


plt.figure(dpi=500, figsize = (s1,s2))
plt.errorbar(sim.RID_rpi, sim.p32_max_4, yerr = sim.δp32_max_4,marker = '.',
             ecolor = 'black', markeredgecolor = 'black', markeredgewidth = .5, 
             ls='none', markersize=8, capsize=3, elinewidth = .5,markerfacecolor='darkorange')
plt.plot(P_bank*1e3, sim.p32_max, color = 'black')
plt.plot(P_bank*1e3, sim.p32_bound, color = 'black', linestyle = '--')
plt.yscale('log')
plt.xscale('log')
plt.ylim(.001,.2)
plt.tick_params(which='both', direction='in')
plt.xticks(xtick_positions, xtick_labels)
plt.grid(axis = 'y', which = 'minor', color = 'black',linewidth = .2)
plt.grid(axis = 'y', which = 'major', color = 'black',linewidth=1)
```
