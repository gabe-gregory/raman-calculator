```python
sim0 = s.sim(ls_order = 8)
sim0.get_f()
sim0.b_pi = 1/3
sim0.b_sp = 1/3

sim1 = s.sim(ls_order = 8)
sim1.get_f()
sim1.b_pi = .33
sim1.b_sp = .373
```

```python
Pb_bank = np.linspace(0.02, 0.2, 20)
sim0.power_scaling(Pb_bank)
sim1.power_scaling(Pb_bank)
```

```python
sim0.analytics(Pb_bank, sim_resonance = sim0.w_num_bank)
sim1.analytics(Pb_bank, sim_resonance = sim1.w_num_bank)
```

```python
w20 = 3*sim0.ω0/2 + sim0.δ02_bank/2 -sim0.δ32_bank/2 
w40 = 3*sim0.ω0/2 + sim0.δ02_bank/2 + sim0.δ04_bank/2 -sim0.δ32_bank/2 -sim0.δ34_bank/2 
w60 = 3*sim0.ω0/2 + sim0.δ02_bank/2 + sim0.δ04_bank/2 + sim0.δ06_bank/2 -sim0.δ32_bank/2 -sim0.δ34_bank/2 -sim0.δ36_bank/2
w80 = 3*sim0.ω0/2 + sim0.δ02_bank/2 + sim0.δ04_bank/2 + sim0.δ06_bank/2 + sim0.δ08_bank/2 -sim0.δ32_bank/2 -sim0.δ34_bank/2 -sim0.δ36_bank/2 -sim0.δ38_bank/2
```

```python
w21 = 3*sim1.ω0/2 + sim1.δ02_bank/2 -sim1.δ32_bank/2 
w41 = 3*sim1.ω0/2 + sim1.δ02_bank/2 + sim1.δ04_bank/2 -sim1.δ32_bank/2 -sim1.δ34_bank/2 
w61 = 3*sim1.ω0/2 + sim1.δ02_bank/2 + sim1.δ04_bank/2 + sim1.δ06_bank/2 -sim1.δ32_bank/2 -sim1.δ34_bank/2 -sim1.δ36_bank/2
w81 = 3*sim1.ω0/2 + sim1.δ02_bank/2 + sim1.δ04_bank/2 + sim1.δ06_bank/2 + sim1.δ08_bank/2 -sim1.δ32_bank/2 -sim1.δ34_bank/2 -sim1.δ36_bank/2 -sim1.δ38_bank/2
```

```python
with plt.style.context(['science','no-latex']):
    plt.figure(dpi = 400, figsize = (4,4))
    # plt.axhline(res_bare/(2*np.pi*1e6), color = 'black', linewidth = 0.1)
    plt.plot(Pb_bank*1e3, w20/(2*np.pi*1e6), label = '2', color = 'black', linestyle = 'dashed')
    plt.plot(Pb_bank*1e3, w40/(2*np.pi*1e6), label = '4', color = 'black', linestyle = 'dashdot')
    plt.plot(Pb_bank*1e3, w60/(2*np.pi*1e6), label = '6', color = 'black', linestyle = 'dotted')
    plt.plot(Pb_bank*1e3, w80/(2*np.pi*1e6), label = '8', color = 'black')
    plt.errorbar(Pb_bank*1e3, sim0.w_num_bank/(2*np.pi*1e6), 
                 sim0.δw_num_bank/(2*np.pi*1e6),label = 'simulation', 
                 ls = 'none', marker = '^', markersize = 2, color = 'magenta')
    #plt.legend(title = 'up to order:')
    plt.ylabel('four-photon resonance (MHz)')
    plt.xlim(0,200)
    plt.xlabel(r'R$_\perp$ power (mW)')
    plt.ylim(3.615, 3.655)
    plt.show()
```
