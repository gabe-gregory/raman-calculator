```python
w2 = 3*sim.ω0/2 + sim.δ02_bank/2 -sim.δ32_bank/2 
w4 = 3*sim.ω0/2 + sim.δ02_bank/2 + sim.δ04_bank/2 -sim.δ32_bank/2 -sim.δ34_bank/2 
w6 = 3*sim.ω0/2 + sim.δ02_bank/2 + sim.δ04_bank/2 + sim.δ06_bank/2 -sim.δ32_bank/2 -sim.δ34_bank/2 -sim.δ36_bank/2
w8 = 3*sim.ω0/2 + sim.δ02_bank/2 + sim.δ04_bank/2 + sim.δ06_bank/2 + sim.δ08_bank/2 -sim.δ32_bank/2 -sim.δ34_bank/2 -sim.δ36_bank/2 -sim.δ38_bank/2
```
