```python
sim = s.sim()
sim.ls_854 = 2*np.pi*500e3
sim.get_f()
```

```python
def two_photon(P_bank, w0b):
    sim.w0b = w0b
    sim.power_scaling(P_bank)
    return sim.Ω_num_bank

def fit_two_photon(x_data, y_data, sigma, p0):
    popt, pcov = curve_fit(two_photon, x_data, y_data, p0 = p0, sigma = sigma)
    error = np.sqrt(np.diag(pcov))
    return popt, error
```

```python
x_data = sim.RID_rpi*1e-3
y_data = np.pi/(sim.tpi_2)
δy_data = np.pi*sim.δtpi_2/(sim.tpi_2**2)

p0 = [34e-6]

params, error = fit_two_photon(x_data, y_data, δy_data, p0)
```

```python
two_photon(x_data, params[0])
```

```python
Ω2_num = sim.Ω_num_bank/(2*np.pi*1e3)
δΩ2_num = sim.δΩ_num_bank/(2*np.pi*1e3)
Ω2_data = np.pi*1e-3/(2*np.pi*sim.tpi_2)
δΩ2_data = np.pi*1e-3*sim.δtpi_2/(2*np.pi*sim.tpi_2**2)
```

```python
def Ω2_func(Pb,A):
    return A*Pb**.5

def Ω2_fit(x_data, y_data, p0_guess):
    popt, pcov = curve_fit(Ω2_func, x_data, y_data, p0 = p0_guess)
    error = np.sqrt(np.diag(pcov))
    return popt, error
```
