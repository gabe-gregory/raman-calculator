```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
Bs = np.linspace(1, 700, 1000)
Is = Bs
Dets = 1/(Bs**.5)

fig, ax1 = plt.subplots(figsize = (4,3), dpi = 200)

color1 = "darkblue"
ax1.plot(Bs, Is, color=color1, label="Is vs Bs")
ax1.set_xlabel("Bs")
ax1.set_ylabel("Is", color=color1)
ax1.tick_params(axis="y", labelcolor=color1)

# right y-axis
ax2 = ax1.twinx()
color2 = "darkred"
ax2.plot(Bs, Dets, color=color2, label="Dets vs Bs")
ax2.set_ylabel("Dets", color=color2)
ax2.tick_params(axis="y", labelcolor=color2)

plt.xscale('log')
ax1.set_yscale('log')
ax2.set_yscale('log')
ax1.set_ylim(1,max(Bs))
ax2.set_ylim(1/max(Bs), 1)
ax1.set_xlim(min(Bs), max(Bs))
ax1.set_xlabel('|B| field factor increase')
ax1.set_ylabel('R  intensity factor increase')
ax2.set_ylabel('P state detuning factor increase')

ax1.set_title('R  intensity / P state detuning compensation')
plt.show()
```
